```python
# Financial Time Series Forecasting
# Author: Angelo Sorte
# Description: Forecast financial trends using time series data

# ------------------------
# Imports
# ------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# ------------------------
# Load dataset
# ------------------------
# Replace 'financial_timeseries.csv' with your dataset
data = pd.read_csv('financial_timeseries.csv')

# Convert date column
data['Date'] = pd.to_datetime(data['Date'])
data.sort_values('Date', inplace=True)

# ------------------------
# Feature engineering
# ------------------------
# Create time index for regression
data['TimeIndex'] = np.arange(len(data))

X = data[['TimeIndex']]
y = data['Value']  # target variable (e.g., stock price)

# ------------------------
# Train model
# ------------------------
model = LinearRegression()
model.fit(X, y)

# ------------------------
# Predict future values
# ------------------------
future_steps = 10
future_index = np.arange(len(data), len(data) + future_steps).reshape(-1, 1)
future_predictions = model.predict(future_index)

# ------------------------
# Visualization
# ------------------------
plt.figure(figsize=(10,6))
plt.plot(data['Date'], y, label='Historical Data')
future_dates = pd.date_range(start=data['Date'].iloc[-1], periods=future_steps+1, freq='D')[1:]
plt.plot(future_dates, future_predictions, label='Forecast', linestyle='--')
plt.legend()
plt.title("Financial Time Series Forecast")
plt.xlabel("Date")
plt.ylabel("Value")
plt.show()
```
